## B. Data Understanding

```python
# Do not modify
try:
  df = pd.read_csv(at.folder_path / 'special_offer.csv')
except Exception as e:
  print(e)
```

### B.1 Explore Dataset

```python
print('Shape:', df.shape)
print('Columns:', list(df.columns))
df.head(5)
```

```python
print(df.dtypes)
print('Missing:')
print(df.isnull().sum())
df.describe(include='all')
```

```python
print(df['type'].value_counts())
print(df['category'].value_counts())
```

```python
dataset_insights = """
The special_offer dataset has 16 rows and 9 columns representing promotional offers.
Columns: special_offer_id, description, discount_pct, type, category, min_quantity, max_quantity, start_date, end_date.
12 missing values only in max_quantity - many offers have no upper quantity limit.
Discount percentages range from 0 to significant clearance discounts.
This dataset is critical as discount exposure is a key product clustering dimension.
"""
```

```python
# Do not modify
print_tile(size='h3', key='dataset_insights', value=dataset_insights)
```

### B.2 Explore Feature of Interest `discount_pct`

```python
print(df['discount_pct'].describe())
print('Unique values:', sorted(df['discount_pct'].unique()))

discount_bar_chart = alt.Chart(df).mark_bar(color='tomato').encode(
    x=alt.X('description:N', sort='-y', title='Offer', axis=alt.Axis(labelAngle=-45)),
    y='discount_pct:Q',
    tooltip=['description','discount_pct','type']
)

discount_histogram = alt.Chart(df).mark_bar().encode(
    alt.X('discount_pct:Q', bin=True, title='Discount Percentage'),
    y='count():Q'
)

alt.hconcat(discount_bar_chart, discount_histogram).properties(title='Discount Percentage Analysis')
```

```python
feature_1_insights = """
discount_pct is the most important column for the clustering use case.
Values range from 0.0 (no discount) up to significant clearance discounts.
Most offers cluster at low discount percentages with a few carrying large discounts.
Will be aggregated per product as avg/max discount - a key clustering feature.
"""
```

```python
# Do not modify
print_tile(size='h3', key='feature_1_insights', value=feature_1_insights)
```

### B.3 Explore Feature of Interest `type` and `category`

```python
print(df['type'].value_counts())
print(df['category'].value_counts())

chart_type_category = alt.Chart(df).mark_bar().encode(
    x=alt.X('type:N', title='Offer Type'),
    y=alt.Y('count():Q', title='Count of Offers'),
    color='category:N',
    tooltip=['type', 'category', 'count():Q']
).properties(width=400, height=300)

chart_category_count = alt.Chart(df).mark_bar().encode(
    x=alt.X('category:N', title='Category'),
    y=alt.Y('count():Q', title='Count of Offers'),
    tooltip=['category', 'count():Q']
).properties(width=300, height=300)

alt.hconcat(chart_type_category, chart_category_count).properties(
    title='Analysis of Offer Types and Categories'
)
```

```python
feature_2_insights = """
The type column describes the offer nature: Volume Discount, Clearance, Seasonal, Excess Inventory.
The category column classifies offers as Customer or Reseller-facing.
Volume Discounts dominate suggesting the retailer incentivises bulk purchasing.
Will be encoded as binary flags per product for the clustering model.
"""
```

```python
# Do not modify
print_tile(size='h3', key='feature_2_insights', value=feature_2_insights)
```

### B.4 Explore Feature of Interest `min_quantity`

```python
print(df['min_quantity'].describe())
print('max_quantity nulls:', df['max_quantity'].isnull().sum())
alt.Chart(df.dropna(subset=['max_quantity'])).mark_point(size=80, color='purple').encode(
    x='min_quantity:Q', y='max_quantity:Q',
    tooltip=['description','min_quantity','max_quantity','discount_pct']
).properties(title='Min vs Max Quantity', width=350, height=250)
```

```python
feature_n_insights = """
min_quantity defines the minimum purchase to activate a discount.
max_quantity has 12 missing values (75%) - most offers are open-ended with no upper limit.
Quantity thresholds reveal volume strategy of the retailer.
For clustering avg min_quantity per product may be a useful engineered feature.
"""
```

```python
# Do not modify
print_tile(size='h3', key='feature_n_insights', value=feature_n_insights)
```
