## B. Data Understanding

```python
# Do not modify
try:
  df = pd.read_csv(at.folder_path / 'special_offer_product.csv')
except Exception as e:
  print(e)
```

### B.1 Explore Dataset

```python
print('Shape:', df.shape)
print('Columns:', list(df.columns))
df.head(10)
```

```python
print(df.dtypes)
print('Missing:', df.isnull().sum())
print('Unique offers:', df['special_offer_id'].nunique())
print('Unique products:', df['product_id'].nunique())
prod_per_offer = df.groupby('special_offer_id')['product_id'].count().reset_index()
prod_per_offer.columns = ['special_offer_id','product_count']
print(prod_per_offer.sort_values('product_count', ascending=False))
```

```python
dataset_insights = """
The special_offer_product dataset is a junction table with 538 rows and 2 columns.
Columns: special_offer_id, product_id - linking offers to products.
No missing values. Covers 15 unique offers and 295 unique products.
Key table for computing per-product discount exposure metrics for clustering.
"""
```

```python
# Do not modify
print_tile(size='h3', key='dataset_insights', value=dataset_insights)
```

### B.2 Explore Feature of Interest `product_id` (offers per product)

```python
offers_pp = df.groupby('product_id')['special_offer_id'].count().reset_index()
offers_pp.columns = ['product_id','num_offers']
print(offers_pp['num_offers'].describe())
print(offers_pp['num_offers'].value_counts().sort_index())
alt.Chart(offers_pp).mark_bar(color='steelblue').encode(
    x=alt.X('num_offers:O', title='Offers per Product'),
    y='count():Q'
).properties(title='How Many Offers Per Product', width=400, height=280)
```

```python
feature_1_insights = """
Aggregating product_id reveals how many offers each product has appeared in.
Most products appear in 1-2 offers while a smaller group appears in 3-5.
Products with many offers may be slow-movers needing frequent discounting.
This count (num_offers per product) is one of the most important clustering features.
"""
```

```python
# Do not modify
print_tile(size='h3', key='feature_1_insights', value=feature_1_insights)
```

### B.3 Explore Feature of Interest `special_offer_id` (products per offer)

```python
print(prod_per_offer.sort_values('product_count', ascending=False))
alt.Chart(prod_per_offer).mark_bar(color='orange').encode(
    x=alt.X('special_offer_id:N', axis=alt.Axis(labelLimit=60)),
    y='product_count:Q'
).properties(title='Products per Special Offer', width=500, height=280).configure_axisX(labelAngle=-45)
```

```python
feature_2_insights = """
Some offers cover nearly all products (broad volume discounts) others are product-specific.
Two types emerge: broad market-wide promotions and targeted product clearances.
Products only in broad discounts behave differently from those in targeted clearance offers.
"""
```

```python
# Do not modify
print_tile(size='h3', key='feature_2_insights', value=feature_2_insights)
```

### B.4 Explore Feature of Interest `Average Discount per Product` (derived)

```python
try:
  so = pd.read_csv(at.folder_path / 'special_offer.csv')
except Exception as e:
  print(e)
df_m = df.merge(so[['special_offer_id','discount_pct']], on='special_offer_id', how='left')
avg_disc = df_m.groupby('product_id')['discount_pct'].mean().reset_index()
avg_disc.columns = ['product_id','avg_discount_pct']
print(avg_disc['avg_discount_pct'].describe())
alt.Chart(avg_disc).mark_bar(color='crimson').encode(
    x=alt.X('avg_discount_pct:Q', bin=alt.Bin(maxbins=20), title='Avg Discount % per Product'),
    y='count():Q'
).properties(title='Avg Discount % per Product', width=400, height=280)
```

```python
feature_n_insights = """
Joining both tables gives avg discount percentage per product.
Most products have very low average discounts as the No Discount offer dominates.
A smaller group has meaningful discounts (5-35%) indicating heavily promoted products.
avg_discount_pct per product is a powerful clustering feature separating premium from promotional products.
"""
```

```python
# Do not modify
print_tile(size='h3', key='feature_n_insights', value=feature_n_insights)
```
