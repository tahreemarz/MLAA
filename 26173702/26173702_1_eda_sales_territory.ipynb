## B. Data Understanding

```python
# Do not modify
try:
  df = pd.read_csv(at.folder_path / 'sales_territory.csv')
except Exception as e:
  print(e)
```

### B.1 Explore Dataset

```python
print('Shape:', df.shape)
print('Columns:', list(df.columns))
df.head(10)
```

```python
print(df.dtypes)
print('Missing values:')
print(df.isnull().sum())
df.describe(include='all')
```

```python
dataset_insights = """
The sales_territory dataset has 4 rows and 8 columns representing geographic sales regions.
Columns: territory_id, name, CountryRegionCode, group, sales_ytd, sales_last_year, cost_ytd, cost_last_year.
No missing values - dataset is completely clean.
cost_ytd and cost_last_year are all 0.0 meaning costs are not tracked at territory level.
This small dataset serves as contextual reference data for geographic cluster interpretation.
"""
```

```python
# Do not modify
print_tile(size='h3', key='dataset_insights', value=dataset_insights)
```

### B.2 Explore Feature of Interest `sales_ytd`

```python
print(df['sales_ytd'].describe())
chart = alt.Chart(df).mark_bar(color='steelblue').encode(
    x=alt.X('name:N', sort='-y', title='Territory'),
    y=alt.Y('sales_ytd:Q', title='Sales YTD ($)'),
    tooltip=['name','sales_ytd','group']
).properties(title='Sales YTD by Territory', width=400, height=280)
chart
```

```python
feature_1_insights = """
sales_ytd represents year-to-date revenue per territory.
Values vary considerably revealing performance disparity between regions.
With only 4 territories statistical inference is very limited.
Provides geographic context for interpreting product clusters.
"""
```

```python
# Do not modify
print_tile(size='h3', key='feature_1_insights', value=feature_1_insights)
```

### B.3 Explore Feature of Interest `sales_last_year`

```python
print(df['sales_last_year'].describe())
df_m = df.melt(id_vars='name', value_vars=['sales_ytd','sales_last_year'], var_name='Period', value_name='Sales')
alt.Chart(df_m).mark_bar().encode(
    x='name:N', y='Sales:Q', color='Period:N', column='Period:N'
).properties(title='YTD vs Last Year', width=150, height=200)
```

```python
feature_2_insights = """
sales_last_year enables year-over-year comparison per territory.
Some territories show growth while others appear flat.
No missing values so no imputation is required.
Growth trends help identify which product segments to prioritise per region.
"""
```

```python
# Do not modify
print_tile(size='h3', key='feature_2_insights', value=feature_2_insights)
```

### B.4 Explore Feature of Interest `group`

```python
print(df['group'].value_counts())
alt.Chart(df).mark_bar(color='orange').encode(
    x='group:N', y='count():Q'
).properties(title='Territories by Group', width=300, height=220)
```

```python
feature_n_insights = """
The group column categorises territories by continental region.
All 4 territories belong to different groups providing geographic diversity.
Being categorical this column will be encoded if used in model features.
Helps interpret which geographic markets each product cluster targets.
"""
```

```python
# Do not modify
print_tile(size='h3', key='feature_n_insights', value=feature_n_insights)
```
