## B. Data Understanding

```python
# Do not modify
try:
  df = pd.read_csv(at.folder_path / 'product_cost_history.csv')
except Exception as e:
  print(e)
```

### B.1 Explore Dataset

```python
print('Shape:', df.shape)
print('Columns:', list(df.columns))
df.head(10)
```

```python
print(df.dtypes)
print('Missing:', df.isnull().sum())
print('Unique products:', df['product_id'].nunique())
df.describe()
```

```python
recs = df.groupby('product_id').size().reset_index(name='n')
print('Cost records per product:')
print(recs['n'].value_counts().sort_index())
```

```python
dataset_insights = """
product_cost_history has 395 rows and 4 columns tracking cost changes per product over time.
Columns: product_id, start_date, end_date, standard_cost.
195 missing values in end_date (~49%) - these are currently active open cost records.
Covers 293 unique products. standard_cost ranges from $0.86 to $2171.29.
Strongly right-skewed distribution suggesting two tiers: accessories and complete bikes.
Central dataset for clustering as cost evolution differentiates product segments.
"""
```

```python
# Do not modify
print_tile(size='h3', key='dataset_insights', value=dataset_insights)
```

### B.2 Explore Feature of Interest `standard_cost`

```python
print(df['standard_cost'].describe())
alt.Chart(df).mark_bar(color='teal').encode(
    x=alt.X('standard_cost:Q', bin=alt.Bin(maxbins=30), title='Standard Cost ($)'),
    y='count():Q'
).properties(title='Distribution of Standard Cost', width=450, height=300)
```

```python
feature_1_insights = """
standard_cost is the manufacturing/procurement cost per product per time period.
Distribution is strongly right-skewed: most products have low costs but premium items are very expensive.
Bimodal nature suggests two natural tiers: low-cost accessories and high-cost complete bikes.
Aggregated avg_cost per product anchors the cost dimension of our clustering feature set.
"""
```

```python
# Do not modify
print_tile(size='h3', key='feature_1_insights', value=feature_1_insights)
```

### B.3 Explore Feature of Interest `Cost Volatility` (derived)

```python
cost_vol = df.groupby('product_id')['standard_cost'].agg(['std','max','min']).reset_index()
cost_vol.columns = ['product_id','cost_std','max_cost','min_cost']
cost_vol['cost_change'] = cost_vol['max_cost'] - cost_vol['min_cost']
cost_vol['cost_std'] = cost_vol['cost_std'].fillna(0)
print(cost_vol['cost_std'].describe())
alt.Chart(cost_vol).mark_bar(color='darkorange').encode(
    x=alt.X('cost_std:Q', bin=alt.Bin(maxbins=25), title='Cost Std Dev ($)'),
    y='count():Q'
).properties(title='Cost Volatility per Product', width=450, height=280)
```

```python
feature_2_insights = """
Cost volatility (std dev per product) captures pricing instability over time.
Most products have zero volatility (single cost record) - stable mature products.
High volatility may indicate supply chain issues or material cost fluctuations.
cost_std and cost_change included as clustering features to differentiate stability profiles.
"""
```

```python
# Do not modify
print_tile(size='h3', key='feature_2_insights', value=feature_2_insights)
```

### B.4 Explore Feature of Interest `end_date` (active records)

```python
df['is_active'] = df['end_date'].isnull().astype(int)
print('Active records (no end_date):', df['is_active'].sum())
print('Closed records:', (1-df['is_active']).sum())
alt.Chart(df).mark_bar().encode(
    x=alt.X('is_active:O', title='Is Active (1=Open)'),
    y='count():Q',
    color=alt.Color('is_active:O', legend=None)
).properties(title='Active vs Closed Cost Records', width=300, height=250)
```

```python
feature_n_insights = """
end_date has 195 missing values (49%) representing currently active cost records.
Missing end_date is NOT a data quality issue - it means the cost is currently valid.
Active records indicate the product is still being manufactured and sold.
A binary flag has_active_cost will be created to distinguish active from discontinued products.
"""
```

```python
# Do not modify
print_tile(size='h3', key='feature_n_insights', value=feature_n_insights)
```
