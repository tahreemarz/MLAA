### B.1 Explore Dataset

```python
# <Student to fill this section>
# Shape, schema and overall structure
print("Shape:", df.shape)
df.info()
df.head()
```

```python
# Missing values, duplicates, uniqueness of the primary key
print("Missing values per column (top 12):")
print(df.isnull().sum().sort_values(ascending=False).head(12))

print(f"\nTotal rows: {len(df)}")
print(f"Unique product_id: {df['product_id'].nunique()}")
print(f"Fully duplicated rows: {df.duplicated().sum()}")
print(f"Rows whose product_id is duplicated: {df['product_id'].duplicated(keep=False).sum()}")

# What does a duplicated product look like?
dup_pid = df[df['product_id'].duplicated(keep=False)]['product_id'].iloc[0]
print(f"\nAll rows for one duplicated product_id ({dup_pid[:8]}...):")
print(df[df['product_id'] == dup_pid].head(3))
```

```python
# <Student to fill this section>
dataset_insights = """
The product dataset has 886 rows and 23 columns covering manufacturing, pricing, dimensions and lifecycle information.

Critical data-quality finding:
- product_id has only 504 unique values across 886 rows. 382 rows are fully duplicated (every column identical). When deduplicated by product_id, every duplicate set has 100% identical attribute values, so the duplication is a load/export artifact rather than conflicting records. The fix is a simple drop_duplicates() in the Preparation notebook.

Other observations:
- discontinued_date is 100% null and provides no information; safe to drop.
- sell_end_date is null for 81% of products, which is the expected behaviour (most products are still actively sold).
- Optional product attributes are heavily missing: size (57%), size_unit_measure_code (64%), style (57%), class (49%), color (48%), product_line (44%). This is because the table mixes finished goods (which have colours, sizes, styles) with raw materials and components (which do not). is_manufactured and is_sellable flag this distinction (only 212 of 504 unique products are both manufactured AND sellable).
- list_price and product_id are 100% populated, which is important because list_price is the anchor of our anomaly use case (we will compare each sales line's unit_price against this value).

Implications for the anomaly-detection use case:
We will join sales_order_detail to product to bring in standard_cost, list_price, product_subcategory_id and class. The duplication issue must be fixed first, otherwise every line item will be inflated 2-6 times. We will also need to filter out products with list_price = 0 (raw materials sold internally) before computing discount-based features.
"""
```

```python
# Do not modify this code
print_tile(size="h3", key='dataset_insights', value=dataset_insights)
```

### B.2 Explore Feature of Interest `\<list_price\>`

```python
# <Student to fill this section>
# Deduplicate first since the same product_id repeats
df_u = df.drop_duplicates(subset='product_id').copy()
print(f"Unique products after dedup: {len(df_u)}")

print("\nlist_price distribution:")
print(df_u['list_price'].describe())

print(f"\nProducts with list_price == 0: {(df_u['list_price'] == 0).sum()}  "
      f"({(df_u['list_price']==0).mean()*100:.1f}%)")
print(f"Products with list_price > 0:  {(df_u['list_price'] > 0).sum()}  "
      f"({(df_u['list_price']>0).mean()*100:.1f}%)")
```

```python
# Histogram of list_price for products that are actually priced
plot_df = df_u[df_u['list_price'] > 0].copy()

hist = alt.Chart(plot_df).mark_bar(color='#4C78A8').encode(
    x=alt.X('list_price:Q',
            bin=alt.Bin(maxbins=40),
            title='List price (USD)'),
    y=alt.Y('count():Q', title='Number of products'),
    tooltip=['count():Q']
).properties(
    title='Distribution of list_price (products with list_price > 0, n=304)',
    width=600, height=300
)
hist
```

```python
# <Student to fill this section>
feature_1_insights = """
list_price is the recommended retail price for each product and is the anchor for our anomaly model: every sales line item's unit_price will be compared against the product's current list_price.

Distribution (after deduplication, 504 unique products):
- Range 0 to 3,578.27 USD, mean 438.67, median 49.99 - heavily right-skewed.
- A clear bimodal pattern emerges: 200 products (~40%) are priced at exactly 0, while the remaining 304 products span 2.29 to 3,578.27.
- Among priced products, ~60% sit below 100 USD (accessories, components) and ~10% sit above 1,500 USD (high-end bikes).

The 200 zero-price products are raw materials and sub-components (frames, gears, washers) that are never sold to end customers - they are consumed in manufacturing. They have list_price = 0 because they are not retail goods.

Implications:
- For the anomaly model we will only retain sales lines whose product has list_price > 0. The 200 zero-priced products generate no end-customer revenue and including them would dilute the discount/margin signal.
- The strong skew suggests we will need either a log transform on price-derived features (unit_price, line_total) or a tree-based model (Isolation Forest) that is scale-invariant.
"""
```

```python
# Do not modify this code
print_tile(size="h3", key='feature_1_insights', value=feature_1_insights)
```

### B.3 Explore Feature of Interest `\<standard_cost and margin\>`

```python
# <Student to fill this section>
# Cost analysis on the deduplicated product set
print("standard_cost summary (deduped):")
print(df_u['standard_cost'].describe())
print(f"\nNull standard_cost: {df_u['standard_cost'].isna().sum()}")
print(f"Zero standard_cost: {(df_u['standard_cost'] == 0).sum()}")

# Compute gross margin where both cost and price are positive
margin_df = df_u[(df_u['list_price'] > 0) & (df_u['standard_cost'].notna())].copy()
margin_df['margin_pct'] = (margin_df['list_price'] - margin_df['standard_cost']) / margin_df['list_price'] * 100

print(f"\nProducts with both list_price > 0 and a non-null cost: {len(margin_df)}")
print("\nMargin% summary:")
print(margin_df['margin_pct'].describe())
print(f"\nProducts priced BELOW cost (list_price < cost): {(margin_df['list_price'] < margin_df['standard_cost']).sum()}")
```

```python
margin_chart = alt.Chart(margin_df).mark_bar(color='#54A24B').encode(
    x=alt.X('margin_pct:Q', bin=alt.Bin(maxbins=25), title='Gross margin (%)'),
    y=alt.Y('count():Q', title='Number of products'),
    tooltip=['count():Q']
).properties(
    title='Distribution of gross margin (n=272 products with cost and price > 0)',
    width=600, height=300
)
margin_chart
```

```python
# <Student to fill this section>
feature_2_insights = """
standard_cost is the company's internal cost of acquiring or producing a unit. Combined with list_price it defines the company's normal margin structure, which is critical for our anomaly use case: a sales line with unit_price BELOW the product's standard_cost is suspicious (potential pricing error, unauthorised deep discount, or fraud).

Distribution (deduplicated):
- 452 of 504 products have a non-null cost. 180 of those are exactly 0 (the raw-material rows that also have list_price = 0).
- Among 272 products with both a positive cost AND a positive list_price, gross margin ranges from 23.0% to 64.2% with median ~45.2%. The company maintains very healthy and consistent margins.
- Importantly, ZERO products in the master file are priced below cost. This means we have a clean baseline: any sales-line unit_price observed below the product's standard_cost is a real anomaly worth flagging, not a normal pattern.

Limitations:
- 52 products have a null standard_cost - if these appear in transactions we will need to either impute (using subcategory median) or exclude them from cost-based features.
- standard_cost is a single snapshot. The product_cost_history table contains time-varying costs, which we may join in the Preparation notebook if a sales line is meaningfully older than the current cost (out of scope for this MVP).
"""
```

```python
# Do not modify this code
print_tile(size="h3", key='feature_2_insights', value=feature_2_insights)
```

### B.4 Explore Feature of Interest `\<product_subcategory_id\>`

```python
# <Student to fill this section>
# Subcategory coverage and how it interacts with class / line
print("product_subcategory_id missingness:")
print(f"  Null:     {df_u['product_subcategory_id'].isna().sum()}")
print(f"  Non-null: {df_u['product_subcategory_id'].notna().sum()}")
print(f"  Unique subcategories used: {df_u['product_subcategory_id'].nunique()}")

print("\nClass distribution (deduped):")
print(df_u['class'].value_counts(dropna=False))

print("\nproduct_line distribution (deduped):")
print(df_u['product_line'].value_counts(dropna=False))

# Are zero-priced products the same as the ones without subcategory?
print("\nlist_price == 0 vs subcategory_id null cross-tab:")
print(pd.crosstab(df_u['list_price'] == 0,
                  df_u['product_subcategory_id'].isna(),
                  rownames=['list_price==0'], colnames=['subcat is null']))
```

```python
# Top subcategories by product count (we'll merge in subcategory names)
sub_df = pd.read_csv(at.folder_path / "product_sub_category.csv")
top_sub = (df_u['product_subcategory_id'].value_counts()
             .reset_index().rename(columns={'count': 'product_count'})
             .merge(sub_df[['product_subcategory_id', 'name']], on='product_subcategory_id', how='left')
             .head(15))

chart = alt.Chart(top_sub).mark_bar(color='#F58518').encode(
    x=alt.X('product_count:Q', title='Number of products'),
    y=alt.Y('name:N', sort='-x', title='Sub-category'),
    tooltip=['name', 'product_count']
).properties(
    title='Top 15 sub-categories by product count',
    width=600, height=350
)
chart
```

```python
# <Student to fill this section>
feature_n_insights = """
product_subcategory_id links each product to one of 37 sub-categories (which themselves roll up to 4 top-level categories: Bikes, Components, Clothing, Accessories). It is the most useful single categorical feature for the anomaly model.

Coverage:
- 295 of 504 unique products (58.5%) have a non-null subcategory. The 209 nulls correspond almost perfectly to the 200 zero-priced raw materials - the cross-tab confirms every zero-priced product has a null subcategory, while only 9 priced products lack a subcategory (a manageable cleaning case).
- The hierarchical class field (L / M / H, likely Low / Medium / High value) is null for ~51% of products and roughly aligns with the price tier within each subcategory. product_line (R, M, T, S) maps to road / mountain / touring / standard.

Implications for the anomaly model:
- product_subcategory_id will become a core categorical feature. With 37 levels we may either keep it as-is for the tree-based Isolation Forest (which handles categoricals once encoded) or aggregate it to the 4 top-level category names for a simpler one-hot encoding.
- The strong correlation between zero list_price and null subcategory confirms that filtering to "sellable, priced products" upstream will simultaneously fix most of the categorical-missing problem.
- class is too sparse to use as a primary feature but could be a secondary signal (high-value products should attract more scrutiny when discounted).
"""
```

```python
# Do not modify this code
print_tile(size="h3", key='feature_n_insights', value=feature_n_insights)
```
