## A. Assess Baseline Model

```python
# DO NOT MODIFY THE CODE IN THIS CELL
# Load data
try:
  X_train = pd.read_csv(at.folder_path / 'X_train.csv')
  y_train = pd.read_csv(at.folder_path / 'y_train.csv')

  X_val = pd.read_csv(at.folder_path / 'X_val.csv')
  y_val = pd.read_csv(at.folder_path / 'y_val.csv')

  X_test = pd.read_csv(at.folder_path / 'X_test.csv')
  y_test = pd.read_csv(at.folder_path / 'y_test.csv')
except Exception as e:
  print(e)
```

### A.1 Generate Predictions with Baseline Model

```python
y_train = y_train.iloc[:, 0] if y_train.shape[1] == 1 else y_train
y_val = y_val.iloc[:, 0] if y_val.shape[1] == 1 else y_val
y_test = y_test.iloc[:, 0] if y_test.shape[1] == 1 else y_test

mean_pred = y_train.mean()
print("Baseline prediction (mean of y_train): $", round(mean_pred, 2))

y_pred_train = np.full(len(y_train), mean_pred)
y_pred_val = np.full(len(y_val), mean_pred)
y_pred_test = np.full(len(y_test), mean_pred)
```

### A.2 Selection of Performance Metrics

> Provide some explanations on why you believe the performance metrics you chose is appropriate


```python
def report(name, y_true, y_pred):
    rmse = np.sqrt(mean_squared_error(y_true, y_pred))
    mae = mean_absolute_error(y_true, y_pred)
    r2 = r2_score(y_true, y_pred)
    print(f"{name:12s} RMSE=${rmse:>10,.2f} MAE=${mae:>9,.2f} R2={r2:>7.4f}")

print("Metric definitions used in this notebook:")
print("RMSE - root mean squared error, in dollars (penalises large errors)")
print("MAE - mean absolute error, in dollars (interpretable typical error)")
print("R2 - proportion of variance explained")
```

```python
performance_metrics_explanations = """
Three metrics are reported together because each tells a different story. To begin with, RMSE penalises large errors more heavily. It is important here because line_total has a long right tail. A $500 mistake on a $5,000 line is much worse for the business than 50 mistakes of $10 each. This is captured by RMSE. Next is MAE, which records the typical error in dollars. This one is easy to explain to a non-technical stakeholder. R2 tells us how much better the model is than just predicting the mean of the training target. A baseline mean-predictor has R2 = 0 by construction and a perfect model has R2 = 1. These were chosen because this is a regression problem with a continuous, skewed dollar amount as the target.
"""
```

```python
# DO NOT MODIFY THE CODE IN THIS CELL
print_tile(size="h3", key='performance_metrics_explanations', value=performance_metrics_explanations)
```

### A.3 Baseline Model Performance

> Provide some explanations on model performance


```python
print("Mean-predictor baseline")
report("Train", y_train, y_pred_train)
report("Val", y_val, y_pred_val)
report("Test", y_test, y_pred_test)
```

```python
baseline_performance_explanations = """
The baseline simply predicts the mean of y_train (approximately $726) for every row. The numbers seen on validation were roughly as follows: RMSE ~ $1,323, MAE ~ $900, R2 ~ 0.0. Here, MAE ~ $900 means that on a typical line our naive prediction is off by about $900, which is far larger than the median line_total which is $50). This is useless in practice. RMSE ~ $1,323 is even worse because it gets pulled up by the big lines. Whereas, R2 ~ 0 confirms the baseline contains no real predictive information. This baseline is the floor that the real regression model must beat and ideally beat by a lot. Any improvement in RMSE or MAE here is a dollar-level improvement that translates directly to better revenue forecasts.
"""
```
