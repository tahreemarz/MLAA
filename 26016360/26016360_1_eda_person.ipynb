## B. Data Understanding

```python
# Do not modify this code
try:
  df = pd.read_csv(at.folder_path / "person.csv")
except Exception as e:
  print(e)
```

### B.1 Explore Dataset

```python
print("Shape:", df.shape)
print()

df.head()
```

```python
df.info()
```

```python
print("Missing values per column:")
print(df.isnull().sum())
```

```python
print("Missing % per column:")
print((df.isnull().mean() * 100).round(1))
```

```python
print("Total rows:", len(df))

print("\nUnique person_ids:", df["person_id"].nunique())

print("\nFull-row duplicates:", df.duplicated().sum())

print("\nDuplicate person_ids:", df["person_id"].duplicated().sum())

print("\nHow many times each person_id appears:")

print(df["person_id"].value_counts().value_counts().sort_index())
```

```python
dataset_insights = """
The person dataset has 25,163 rows and 10 columns, one per person (customers, employees, contacts). Even though person_id should be unique, 5,191 rows are exact duplicates across all 10 columns. Dropping duplicates on person_id leaves 19,972 unique people. High missingness in optional name fields (additional_contact_info - 99.96%, suffix - 99.7%, title - 94.9%, middle_name - 42.7%). Moreover, name_style and email_promotion are stored as floats but are really categorical codes. Lastly, person_type has 7 categories with a few missing rows.
"""
```

```python
# Do not modify this code
print_tile(size="h3", key='dataset_insights', value=dataset_insights)
```

### B.2 Explore Feature of Interest `person_type`

```python
# Counting each person type
pt_counts = df["person_type"].value_counts(dropna=False).reset_index()
pt_counts.columns = ["person_type", "count"]
print(pt_counts)
```

```python
alt.Chart(pt_counts).mark_bar().encode(
    x=alt.X("person_type:N", sort="-y"),
    y="count:Q",
    tooltip=["person_type","count"]
).properties(width=400, height=300, title="Distribution of person_type")
```

```python
feature_1_insights = """
person_type tells us what role each person plays. The values found are as follows: IN (Individual/Store Contact) - 23229, SC (Store Contact) - 955, GC (General Contact) - 371, EM (Employee) - 359, VC (Vendor Contact) - 192, SP (Sales Person) - 22, NaN (missing) - 35. From the looks of it, the dataset is dominated by IN. The 35 missing values are a small issue but easy to fix by either dropping or filling with the mode. For modelling work that needs person_type, it would be heavily imbalanced.
"""
```

```python
# Do not modify this code
print_tile(size="h3", key='feature_1_insights', value=feature_1_insights)
```

### B.3 Explore Feature of Interest `email_promotion`

```python
ep_counts = df["email_promotion"].value_counts(dropna=False).reset_index()
ep_counts.columns = ["email_promotion", "count"]
print(ep_counts)
```

```python
alt.Chart(ep_counts).mark_bar().encode(
    x="email_promotion:N",
    y="count:Q",
    tooltip=["email_promotion","count"]
).properties(width=350, height=300, title="email_promotion preferences")
```

```python
feature_2_insights = """
email_promotion shows whether the person opts in to marketing emails. Here, 0 = no emails, 1 = Adventure Works emails only, 2 = Adventure Works + partner emails. About 210 rows are missing this value. The variable is stored as float64 but is really a small categorical (3 levels). The vast majority of people have opted out (value 0), so as a marketing feature it has low variability.
"""
```

```python
# Do not modify this code
print_tile(size="h3", key='feature_2_insights', value=feature_2_insights)
```

### B.4 Explore Feature of Interest `title` / `middle_name`

```python
print("title value counts (non-null):")

print()
print(df["title"].value_counts(dropna=False))

print("\nmiddle_name missing %:", round(df["middle_name"].isnull().mean()*100,1))
```

```python
miss = (df.isnull().mean()*100).round(1).reset_index()
miss.columns = ["column", "missing_pct"]
alt.Chart(miss).mark_bar().encode(
    x=alt.X("column:N", sort="-y"),
    y="missing_pct:Q",
    tooltip=["column","missing_pct"]
).properties(width=500, height=300, title="Missing % per column")
```

```python
feature_n_insights = """
Columns title, suffix, additional_contact_info and middle_name all have very high missing rates which is between 43% to 99%. These are essentially "extra/optional" name fields and would not add much value as features. For any modelling work, the easiest path is to drop additional_contact_info and suffix outright and treat title and middle_name as "has_title"/"has_middle_name" binary flags if needed. Overall, the person table is mostly useful as a lookup/dimension table to join customer information into transactions; it does not have strong predictive features on its own.
"""
```

```python
# Do not modify this code
print_tile(size="h3", key='feature_n_insights', value=feature_n_insights)
```
