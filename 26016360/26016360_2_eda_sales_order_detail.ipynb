### B.2 Explore Feature of Interest `line_total`

```python
print("line_total summary:")
print(df["line_total"].describe())
```

```python
print("Percentiles:")
for p in [50, 75, 90, 95, 99]:
    print(f"  {p}th = ${df['line_total'].quantile(p/100):,.2f}")
```

```python
import numpy as np

alt.data_transformers.disable_max_rows()

plot_df = df.copy()
plot_df["log_line_total"] = np.log1p(plot_df["line_total"])

h1 = alt.Chart(plot_df).mark_bar().encode(
    x=alt.X("line_total:Q", bin=alt.Bin(maxbins=50)),
    y="count()"
).properties(width=350, height=250, title="line_total (raw)")

h2 = alt.Chart(plot_df).mark_bar().encode(
    x=alt.X("log_line_total:Q", bin=alt.Bin(maxbins=50)),
    y="count()"
).properties(width=350, height=250, title="log(1 + line_total)")

h1 | h2
```

```python
feature_1_insights = """
line_total is the dollar amount of each order line and will be the target for the regression model. The key points are as follows: Range is from $1.37 (min) to $23,668 (max), mean is ~$724 and median is ~$50. So, half the lines are under $50 but a small tail of big lines pulls the mean way up. Therefore, the raw distribution is extremely right-skewed. Then, log(1+line_total) is much more symmetric which is useful for linear models. This skew also means I need a metric that is robust to large values (e.g. MAE alongside RMSE) and a baseline that is more sensible than just predicting the mean.
"""
```

```python
# Do not modify this code
print_tile(size="h3", key='feature_1_insights', value=feature_1_insights)
```

### B.3 Explore Feature of Interest `order_quantity`

```python
q_counts = df["order_quantity"].value_counts().sort_index().reset_index()
q_counts.columns = ["order_quantity","count"]
print(q_counts.head(15))

print("\nMax order_quantity:", df["order_quantity"].max())
print("% of lines with quantity == 1:", round((df["order_quantity"]==1).mean()*100,1))
```

```python
alt.Chart(q_counts.head(15)).mark_bar().encode(
    x="order_quantity:O",
    y="count:Q",
    tooltip=["order_quantity","count"]
).properties(width=500, height=300, title="order_quantity (top 15 values)")
```

```python
feature_2_insights = """
order_quantity is the number of units sold on the line. It ranges from 1 to 36 which are whole numbers. About 80.5% of lines have quantity = 1, dropping off quickly after that. A small number of bulk orders (>= 10) exist because most orders are single-unit, order_quantity has limited variation but still matters. When it is high, it directly multiplies the dollar value of the line. I will keep it as a numeric feature and consider an extra "is_bulk" flag.
"""
```

```python
# Do not modify this code
print_tile(size="h3", key='feature_2_insights', value=feature_2_insights)
```

### B.4 Explore Feature of Interest `special_offer_id` / `unit_price_discount`

```python
print("unit_price_discount value counts:")
print(df["unit_price_discount"].value_counts().sort_index())
```

```python
print("special_offer_id value counts:")
print(df["special_offer_id"].value_counts())
```

```python
avg_by_offer = df.groupby("special_offer_id")["line_total"].mean().reset_index()
avg_by_offer.columns = ["special_offer_id","avg_line_total"]
alt.Chart(avg_by_offer).mark_bar().encode(
    x=alt.X("special_offer_id:N", sort="-y"),
    y="avg_line_total:Q",
    tooltip=["special_offer_id","avg_line_total"]
).properties(width=550, height=300, title="Average line_total by special_offer_id")
```

```python
feature_n_insights = """
special_offer_id has 10 distinct values. One offer which dominates by far is the "no discount" default offer, accounting for the 41,253 lines where unit_price_discount = 0. The other 9 offers correspond to respectively 2%, 5%, 10%, 15%, 20%, 30%, 40% discounts and account for ~2% of lines. The average line_total varies meaningfully by offer, so special_offer_id is a useful categorical feature. With only 10 levels, it is safe to one-hot encode. Since unit_price_discount is itself derived from special_offer_id, I will not include both as features because they carry the same info.
"""
```

```python
# Do not modify this code
print_tile(size="h3", key='feature_n_insights', value=feature_n_insights)
```
