### 0.b Import Packages

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.dummy import DummyClassifier
from sklearn.metrics import accuracy_score, f1_score, confusion_matrix, ConfusionMatrixDisplay
```

---
## A. Assess Baseline Model

```python
# DO NOT MODIFY THE CODE IN THIS CELL
# Load data
try:
  X_train = pd.read_csv(at.folder_path / 'X_train.csv')
  y_train = pd.read_csv(at.folder_path / 'y_train.csv')

  X_val = pd.read_csv(at.folder_path / 'X_val.csv')
  y_val = pd.read_csv(at.folder_path / 'y_val.csv')

  X_test = pd.read_csv(at.folder_path / 'X_test.csv')
  y_test = pd.read_csv(at.folder_path / 'y_test.csv')
except Exception as e:
  print(e)
```

### A.1 Generate Predictions with Baseline Model

```python
baseline = DummyClassifier(strategy="most_frequent")
baseline.fit(X_train, y_train)

y_pred_val = baseline.predict(X_val)

print("Baseline predictions (first 10):", y_pred_val[:10])
print("Most frequent class:", y_train["repeat_buyer"].mode()[0])
```

### A.2 Selection of Performance Metrics


```python
acc = accuracy_score(y_val, y_pred_val)
f1 = f1_score(y_val, y_pred_val, zero_division=0)

print(f"Accuracy:  {acc:.4f}")
print(f"F1 Score:  {f1:.4f}")

fig, ax = plt.subplots(figsize=(5, 4))
ConfusionMatrixDisplay(confusion_matrix(y_val, y_pred_val)).plot(ax=ax)
plt.title("Baseline Confusion Matrix (Validation)")
plt.tight_layout()
plt.show()
```

```python
performance_metrics_explanations = """
Accuracy and F1 score were selected as the primary evaluation metrics. Accuracy gives a straightforward measure of overall correct predictions, while F1 score is more informative here because it accounts for both precision and recall - in a customer retention context, both false positives (wasting marketing spend on non-repeat buyers) and false negatives (missing actual repeat buyers) carry a real business cost, so optimising for F1 is more appropriate than accuracy alone. A confusion matrix is also included to visualise where the model is making errors.
"""
```

```python
# DO NOT MODIFY THE CODE IN THIS CELL
print_tile(size="h3", key='performance_metrics_explanations', value=performance_metrics_explanations)
```

### A.3 Baseline Model Performance

> Provide some explanations on model performance


```python
print(f"Validation Accuracy: {acc:.4f}")
print(f"Validation F1 Score: {f1:.4f}")
print("\nClass distribution in validation set:")
print(y_val["repeat_buyer"].value_counts())
```

```python
baseline_performance_explanations = """
The DummyClassifier using the most_frequent strategy simply predicts the majority class (repeat_buyer = 0) for every observation, achieving an accuracy of around 59% which reflects the class distribution rather than any learned pattern. The F1 score is 0.0 because the model never predicts the positive class, meaning it has zero recall for repeat buyers - this establishes a clear lower bound that any trained classifier must substantially exceed to be considered useful.
"""
```
