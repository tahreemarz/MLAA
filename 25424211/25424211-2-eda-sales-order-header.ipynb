### B.1 Explore Dataset

```python
print(f"Shape: {df.shape}")
print("\nData types:")
print(df.dtypes)
print("\nMissing values:")
print(df.isnull().sum())
df.head()
```

```python
dataset_insights = """
The sales_order_header dataset contains 31,465 rows and 17 columns covering orders placed between May 2011 and June 2014. Two columns have notable missingness: sales_person_id is null for 87.9% of rows and currency_rate_id is null for 55.6%, both of which reflect structural patterns in the data rather than data entry errors - online orders (the majority) do not have an assigned salesperson, and domestic orders do not require a currency conversion rate.
"""
```

## B.2 Explore Feature of Interest `online_order_flag`

```python
online_counts = df["online_order_flag"].value_counts().reset_index()
online_counts.columns = ["online_order_flag", "count"]
online_counts["label"] = online_counts["online_order_flag"].map({1.0: "Online", 0.0: "In-Store"})

plt.figure(figsize=(6, 4))
sns.barplot(data=online_counts, x="label", y="count", palette="Blues_d")
plt.title("Online vs In-Store Orders")
plt.xlabel("Order Channel")
plt.ylabel("Number of Orders")
plt.tight_layout()
plt.show()

print(online_counts[["label","count"]])
```

```python
feature_1_insights = """
The online_order_flag reveals that 87.9% of all orders (27,659 out of 31,465) were placed online, with only 12.1% coming through in-store channels - this strong skew towards online purchasing is an important contextual factor for any customer behaviour model, as the channel itself may be correlated with order size, product type, and discount sensitivity.
"""
```

```python
# Do not modify this code
print_tile(size="h3", key='feature_1_insights', value=feature_1_insights)
```

## B.3 Explore Feature of Interest `total_due`

```python
plt.figure(figsize=(8, 4))
sns.histplot(df[df["total_due"] < 10000]["total_due"], bins=50, color="steelblue")
plt.title("Distribution of Order Value - total_due (capped at $10,000)")
plt.xlabel("Total Due ($)")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()

print(df["total_due"].describe())
```

```python
feature_2_insights = """
The total_due distribution is heavily right-skewed, with a median order value of around $865 compared to a mean of $3,916, indicating that while most orders are relatively modest in value, a small number of very high-value orders (up to $187,487) are pulling the average up considerably - this kind of dispersion will be important to account for during feature engineering and may suggest the need for log transformation before modelling.
"""
```

```python
# Do not modify this code
print_tile(size="h3", key='feature_2_insights', value=feature_2_insights)
```

## B.4 Explore Feature of Interest `order_date`

```python
df["order_date"] = pd.to_datetime(df["order_date"])
monthly_orders = df.groupby(df["order_date"].dt.to_period("M")).size().reset_index()
monthly_orders.columns = ["month", "order_count"]
monthly_orders["month"] = monthly_orders["month"].astype(str)

plt.figure(figsize=(12, 4))
plt.plot(monthly_orders["month"], monthly_orders["order_count"], color="steelblue", marker="o", markersize=3)
plt.title("Monthly Order Volume Over Time")
plt.xlabel("Month")
plt.ylabel("Number of Orders")
plt.xticks(monthly_orders["month"][::6], rotation=45)
plt.tight_layout()
plt.show()
```

```python
feature_n_insights = """
The order_date column spans from May 2011 to June 2014 and shows a clear upward trend in monthly order volume over time, growing from around 200-250 orders per month in 2011-2012 to over 2,000 per month by early 2014 - this rapid growth suggests the business was scaling significantly during this period, and any time-based features derived from order_date could be valuable predictors in a customer behaviour classification model.
"""
```

```python
# Do not modify this code
print_tile(size="h3", key='feature_n_insights', value=feature_n_insights)
```
